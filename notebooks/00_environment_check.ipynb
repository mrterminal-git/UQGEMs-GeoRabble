# 00 - Environment check

Verify the dedicated `uq-gis` environment before processing project data. Restart the kernel and run all cells when validating the setup.

```python
import sys
from pathlib import Path

project_root = Path.cwd().resolve()
while project_root != project_root.parent and not (project_root / 'pyproject.toml').exists():
    project_root = project_root.parent

if not (project_root / 'pyproject.toml').exists():
    raise RuntimeError('Could not locate the project root containing pyproject.toml')

src_path = str(project_root / 'src')
if src_path not in sys.path:
    sys.path.insert(0, src_path)

project_root
```

```python
from IPython.display import JSON, display

from uqgems.environment import (
    collect_environment,
    missing_commands,
    missing_modules,
    write_report,
)

report = collect_environment()
report_path = write_report(
    report, project_root / 'reports' / 'tables' / 'environment_report.json'
)
display(JSON(report, expanded=False))
print(f'Report written to: {report_path}')
```

```python
modules_missing = missing_modules(report)
commands_missing = missing_commands(report)
if modules_missing or commands_missing:
    raise RuntimeError(
        f'Missing modules: {modules_missing}; missing commands: {commands_missing}'
    )

assert report['conda']['default_env'] == 'uq-gis', report['conda']
print('Environment check passed: all required modules and commands are available in uq-gis.')
```
